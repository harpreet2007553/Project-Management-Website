```python
import numpy as np
class Dense:
    def __init__(self, inputs_dim, output_dim):
        self.w = np.random.randn(inputs_dim, output_dim) * 0.01
        self.b = np.zeros((1,output_dim))
    def forward(self, x):
        self.X = x
        return np.dot(x, self.w) + self.b
    def backward(self, dz):
        self.dw = np.dot(self.X.T, dz)
        self.db = np.sum(dz, axis = 0, keepdims=True)
        self.dx = np.dot(dz, self.w.T)
        return self.dw,self.db,self.dx
```

```python
layer = Dense(4,3)
x = np.random.randn(10,4)
z = layer.forward(x)
```

```python
L = np.sum(z)
dz = np.ones_like(z)
dw,db,dx = layer.backward(dz)
```

```python
def numerical_gradient_check_w(layer, X, epsilon):
  numerical_w = np.empty(layer.w.shape)
  for i in np.ndindex(layer.w.shape):
    org_value = layer.w[i]
    layer.w[i] = org_value + epsilon
    z = layer.forward(X)
    L_plus = np.sum(z)
    layer.w[i] = org_value - epsilon
    L_minus = np.sum(layer.forward(X))
    numerical_w[i] = (L_plus - L_minus) / (2 * epsilon)
    layer.w[i] = org_value
  return numerical_w
```

```python
numerical_dw = numerical_gradient_check_w(layer, x, epsilon = 1e-5)
```

```python
relative_error = np.linalg.norm(numerical_dw - layer.dw) / np.linalg.norm(numerical_dw + layer.dw)
print(relative_error)
```

```python
def numerical_gradient_check_b(layer, X, epsilon):
  numerical_b = np.empty(layer.b.shape)
  for i in np.ndindex(layer.b.shape):
    org_value = layer.b[i]
    layer.b[i] = org_value + epsilon
    z = layer.forward(X)
    L_plus = np.sum(z)
    layer.b[i] = org_value - epsilon
    L_minus = np.sum(layer.forward(X))
    numerical_b[i] = (L_plus - L_minus) / (2 * epsilon)
    layer.b[i] = org_value
  return numerical_b
```

```python
numerical_db = numerical_gradient_check_b(layer,x, epsilon = 1e-5)
relative_error = np.linalg.norm(numerical_db - layer.db) / np.linalg.norm(numerical_db + layer.db)
print(relative_error)
```

```python
class ReLU:
  def forward(self, x):
    self.X = x
    return np.maximum(0,x)
  def backward(self, dA):
    mask = np.where(self.X <= 0, 0, 1)
    return mask*dA
```

```python
layer_relu = ReLU()
A = layer_relu.forward(x)
dZ = layer_relu.backward(np.ones_like(A))
```

```python
def numerical_gradient_check_A(layer, X, epsilon):
  numerical_X = np.empty(layer.X.shape)
  for i in np.ndindex(layer.X.shape):
    org_value = X[i]
    X[i] = org_value + epsilon
    z = layer.forward(X)
    L_plus = np.sum(z)
    X[i] = org_value - epsilon
    L_minus = np.sum(layer.forward(X))
    numerical_X[i] = (L_plus - L_minus) / (2 * epsilon)
    X[i] = org_value
  return numerical_X
```

```python
numerical_dX = numerical_gradient_check_A(layer_relu, x, epsilon = 1e-5)
relative_error = np.linalg.norm(numerical_dX - dZ) / np.linalg.norm(numerical_dX + dZ)
print(relative_error)
```

```python
import math
class sigmoid:
  def forward(self, x):
    self.X = x
    self.A = 1/(1 + np.exp(-x))
    return self.A
  def backward(self, dA):
    dz = dA*self.A*(1-self.A)
    return dz
```

```python
sig = sigmoid()
A = sig.forward(x)
dZ = sig.backward(np.ones_like(A))
def numerical_gradient_check_S(layer, X, epsilon):
  numerical_X = np.empty(layer.X.shape)
  for i in np.ndindex(layer.X.shape):
    org_value = X[i]
    X[i] = org_value + epsilon
    z = layer.forward(X)
    L_plus = np.sum(z)
    X[i] = org_value - epsilon
    L_minus = np.sum(layer.forward(X))
    numerical_X[i] = (L_plus - L_minus) / (2 * epsilon)
    X[i] = org_value
  return numerical_X
numerical_dX = numerical_gradient_check_S(sig, x, epsilon = 1e-5)
relative_error = np.linalg.norm(numerical_dX - dZ) / np.linalg.norm(numerical_dX + dZ)
print(relative_error)
```

```python
class tanh:
  def forward(self, x):
    self.X = x
    self.A = np.tanh(x)
    return self.A
  def backward(self, dA):
    dz = dA*(1 - self.A**2)
    return dz
```

```python
tan = tanh()
A = tan.forward(x)
dZ = tan.backward(np.ones_like(A))
def numerical_gradient_check_S(layer, X, epsilon):
  numerical_X = np.empty(layer.X.shape)
  for i in np.ndindex(layer.X.shape):
    org_value = X[i]
    X[i] = org_value + epsilon
    z = layer.forward(X)
    L_plus = np.sum(z)
    X[i] = org_value - epsilon
    L_minus = np.sum(layer.forward(X))
    numerical_X[i] = (L_plus - L_minus) / (2 * epsilon)
    X[i] = org_value
  return numerical_X
numerical_dX = numerical_gradient_check_S(tan, x, epsilon = 1e-5)
relative_error = np.linalg.norm(numerical_dX - dZ) / np.linalg.norm(numerical_dX + dZ)
print(relative_error)
```

```python
class elu:
  def forward(self, x, alpha):
    self.X = x
    self.alpha = alpha
    self.A = np.where(x>0, x, self.alpha*(np.exp(x)-1))
    return self.A
  def backward(self, dA):
    der = np.where(self.X > 0, 1, self.alpha * np.exp(self.X))
    dz = dA*der
    return dz
```

```python
e = elu()
A = e.forward(x, 0.01)
dA = np.ones_like(A)
dZ = e.backward(dA)
def numerical_gradient_check_elu(layer, X, epsilon):
  numerical_X = np.empty(layer.X.shape)
  for i in np.ndindex(layer.X.shape):
    org_value = X[i]
    X[i] = org_value + epsilon
    z = layer.forward(X, alpha = 0.01)
    L_plus = np.sum(z)
    X[i] = org_value - epsilon
    L_minus = np.sum(layer.forward(X, alpha = 0.01))
    numerical_X[i] = (L_plus - L_minus) / (2 * epsilon)
    X[i] = org_value
  return numerical_X
numerical_dX = numerical_gradient_check_elu(e, x, epsilon = 1e-5)
relative_error = np.linalg.norm(numerical_dX - dZ) / np.linalg.norm(numerical_dX + dZ)
print(relative_error)
```

```python
class gelu:
  def forward(self, x):
    self.X = x
    self.u = np.sqrt(2/np.pi) * (x + 0.044715 * x**3)
    self.A = 0.5*x*(1 + np.tanh(self.u))
    return self.A
  def backward(self, dA):
    g = 1 + np.tanh(self.u)
    f = 0.5*self.X
    dg = (1 - np.tanh(self.u)**2) * np.sqrt(2/np.pi) * (1 + 0.134145 * (self.X**2))
    df = 0.5
    dz = df*g + f*dg
    return dA*dz
```

```python
g = gelu()
A = g.forward(x)
dA = np.ones_like(A)
dZ = g.backward(dA)
def numerical_gradient_check_gelu(layer, X, epsilon):
  numerical_X = np.empty(layer.X.shape)
  for i in np.ndindex(layer.X.shape):
    org_value = X[i]
    X[i] = org_value + epsilon
    z = layer.forward(X)
    L_plus = np.sum(z)
    X[i] = org_value - epsilon
    L_minus = np.sum(layer.forward(X))
    numerical_X[i] = (L_plus - L_minus) / (2 * epsilon)
    X[i] = org_value
  return numerical_X
numerical_dX = numerical_gradient_check_gelu(g, x, epsilon = 1e-5)
relative_error = np.linalg.norm(numerical_dX - dZ) / np.linalg.norm(numerical_dX + dZ)
print(relative_error)
```
